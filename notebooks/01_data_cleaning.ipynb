```python
import pandas as pd
import numpy as np

# Load the dataset from the data folder
df = pd.read_csv('../data/rideshare_kaggle.csv')

# Check the total number of rows and columns (Shape)
print("Data Shape:", df.shape)

# Display the first 5 rows of the dataset to verify contents
df.head()
```

```python
# Convert the 'datetime' column from object (string) to actual datetime format
df['datetime'] = pd.to_datetime(df['datetime'])

# Check the data type of the datetime column to confirm conversion
print("Data type of datetime column:", df['datetime'].dtype)
```

```python
# Filter dataset to include only November (month == 11) data to optimize memory
df_filtered = df[df['month'] == 11].copy()

# Print the final row count of the filtered data
print("Total rows for November:", df_filtered.shape[0])
```

```python
# Save the cleaned and filtered dataset as a new CSV file
df_filtered.to_csv('../data/cleaned_rideshare.csv', index=False)

print("Success: Cleaned dataset saved to data folder!")
```
